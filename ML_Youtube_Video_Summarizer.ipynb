```python
def get_video_id(url_link):
  return url_link.split("watch?v=")[-1]
```

```python
video_id = get_video_id("https://www.youtube.com/watch?v=7owOKKgT0ak")
```

```python
from youtube_transcript_api import YouTubeTranscriptApi
```

```python
ytt_api = YouTubeTranscriptApi()

transcript = ytt_api.fetch(video_id)
```

```python
transcript_joined = " ".join(
    snippet.text for snippet in transcript.snippets
)

print(transcript_joined)
```

```python
import re

def clean_transcript(text):
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

cleaned_transcript = clean_transcript(transcript_joined)

print(cleaned_transcript)
```
